```python
# 配置 matplotlib 中文显示
from matplotlib import font_manager
import matplotlib.pyplot as plt
import numpy as np

font_path = "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc"
font_manager.fontManager.addfont(font_path)
font_name = font_manager.FontProperties(fname=font_path).get_name()
plt.rcParams["font.family"] = font_name
plt.rcParams["axes.unicode_minus"] = False
print(f"中文字体配置完成: {font_name}")
```

## 实验1：W7 知识地图

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 10); ax.set_ylim(0, 8); ax.axis('off')

nodes = {"行为设计\nDay1":(2,7), "记忆系统\nDay2":(5,7), "工作流\nDay3":(8,7),
         "多Agent\nDay4":(5,4), "评估体系\nDay5":(5,1.5), "数字员工\n原型":(5,5.5)}
colors = ["#4CAF50", "#2196F3", "#FF9800", "#9C27B0", "#E91E63", "#FF6B6B"]

for (label, (x, y)), color in zip(nodes.items(), colors):
    ax.add_patch(plt.Circle((x, y), 0.7, color=color, alpha=0.3))
    ax.text(x, y, label, ha='center', va='center', fontsize=10, fontweight='bold')

connections = [("行为设计\nDay1","数字员工\n原型"), ("记忆系统\nDay2","数字员工\n原型"),
    ("工作流\nDay3","数字员工\n原型"), ("多Agent\nDay4","数字员工\n原型"),
    ("评估体系\nDay5","数字员工\n原型"), ("行为设计\nDay1","记忆系统\nDay2"),
    ("记忆系统\nDay2","工作流\nDay3"), ("多Agent\nDay4","评估体系\nDay5")]
for s, t in connections:
    sx, sy = nodes[s]; tx, ty = nodes[t]
    ax.annotate("", xy=(tx, ty), xytext=(sx, sy),
                arrowprops=dict(arrowstyle="->", color="gray", lw=1.5, connectionstyle="arc3,rad=0.1"))

ax.set_title("W7 数字员工架构深化 — 知识地图", fontsize=14)
plt.tight_layout()
plt.show()
```

## 实验2：综合能力评估 — 完整数字员工画像

```python
from math import pi

dims = ["行为设计", "记忆系统", "工作流", "多Agent协作", "质量评估", "系统整合"]
N = len(dims)
angles = [n/N*2*pi for n in range(N)]
angles += angles[:1]

early = [7, 3, 2, 1, 2, 1, 7]
final = [9, 8, 8, 7, 8, 8, 9]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
ax.plot(angles, early, 'o-', color="#FF6B6B", linewidth=1.5, label="Day1 结束")
ax.fill(angles, early, alpha=0.1, color="#FF6B6B")
ax.plot(angles, final, '^-', color="#4CAF50", linewidth=2, label="Day7 复习后")
ax.fill(angles, final, alpha=0.15, color="#4CAF50")
ax.set_xticks(angles[:-1]); ax.set_xticklabels(dims, fontsize=11)
ax.set_ylim(0, 10)
ax.set_title("W7 学习成长雷达图", fontsize=14)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1))
plt.tight_layout()
plt.show()
```

## 实验3：每日学习投入与掌握度

```python
days = ["Day1", "Day2", "Day3", "Day4", "Day5", "Day6", "Day7"]
study_hours = [2.5, 3.0, 3.5, 3.0, 2.5, 4.0, 2.0]
mastery = [65, 72, 78, 80, 85, 92, 95]

fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()

ax1.bar(days, study_hours, color="#4ECDC4", alpha=0.7, label="学习时长(h)")
ax1.set_ylabel("学习时长 (h)", color="#4ECDC4")
ax1.set_ylim(0, 5)

ax2.plot(days, mastery, 'o-', color="#E91E63", linewidth=2.5, markersize=8, label="掌握度(%)")
ax2.set_ylabel("掌握度 (%)", color="#E91E63")
ax2.set_ylim(50, 100)

ax1.set_title("W7 每日学习投入与掌握度增长")
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1+lines2, labels1+labels2, loc='upper left')
plt.tight_layout()
plt.show()
print("🎉 W7 完成！你已掌握数字员工架构的核心知识体系。")
```
