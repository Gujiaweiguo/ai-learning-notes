## 🔍 CNN层结构可视化

CNN就像一个特征提取流水线，每一层都有特定的工作内容：

📥 **输入层**：原始图像数据
🔄 **卷积层**：用滤波器滑动提取特征
⚡ **激活函数**：引入非线性，让网络能学习复杂模式
📉 **池化层**：降维，保留重要特征
🔗 **全连接层**：将特征映射到最终分类结果

让我们用一个简单的例子来理解这个过程！

```python
# 加载示例图像
image = load_sample_image('china.jpg') / 255.0
print(f"图像尺寸: {image.shape}")

# 显示原图
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("原始图像")
plt.axis('off')

# 转换为灰度图便于卷积演示
gray_image = cv2.cvtColor((image * 255).astype(np.uint8), cv2.COLOR_RGB2GRAY) / 255.0
plt.subplot(1, 2, 2)
plt.imshow(gray_image, cmap='gray')
plt.title("灰度图像")
plt.axis('off')

plt.tight_layout()
plt.show()
```
