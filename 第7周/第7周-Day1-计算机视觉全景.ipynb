## 🔍 各任务详细解析

### 1. 图像分类（Classification）
**特点**：最基础的任务，回答"这是什么？"
**技术**：CNN、ResNet、ViT
**应用**：图片分类、人脸识别、场景识别

```python
# 图像分类的可视化
fig, ax = plt.subplots(1, 1, figsize=(10, 6))
ax.set_title('图像分类流程', fontsize=14, fontweight='bold')

# 创建流程图
steps = ['输入图片', '特征提取', '分类', '输出结果']
y_pos = np.arange(len(steps))
colors = ['lightblue', 'lightgreen', 'lightcoral', 'lightyellow']

for i, (step, color) in enumerate(zip(steps, colors)):
    ax.barh(i, 1, color=color, alpha=0.7, height=0.6)
    ax.text(0.5, i, step, ha='center', va='center', fontweight='bold', fontsize=12)

ax.set_xlim(0, 1)
ax.set_ylim(-0.5, len(steps)-0.5)
ax.set_yticks(y_pos)
ax.set_yticklabels(steps)
ax.set_xlabel('处理流程')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 2. 目标检测（Detection）
**特点**：回答"哪里有什么？"
**技术**：YOLO、Faster R-CNN、RT-DETR
**输出**：边界框 + 类别标签 + 置信度

```python
# 模拟目标检测可视化
fig, ax = plt.subplots(1, 1, figsize=(10, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 8)
ax.set_aspect('equal')
ax.set_title('目标检测示例', fontsize=14, fontweight='bold')

# 绘制一些随机边界框
objects = [
    {'name': '猫', 'bbox': [2, 3, 4, 5], 'color': 'red'},
    {'name': '狗', 'bbox': [5, 1, 7, 3], 'color': 'blue'},
    {'name': '汽车', 'bbox': [1, 5, 3, 7], 'color': 'green'},
    {'name': '行人', 'bbox': [6, 4, 8, 6], 'color': 'orange'}
]

for obj in objects:
    x1, y1, x2, y2 = obj['bbox']
    width = x2 - x1
    height = y2 - y1
    
    # 绘制边界框
    rect = patches.Rectangle((x1, y1), width, height, 
                           linewidth=2, edgecolor=obj['color'], 
                           facecolor='none', alpha=0.8)
    ax.add_patch(rect)
    
    # 添加标签
    ax.text(x1, y1-0.3, f'{obj["name"]}', 
            fontsize=12, fontweight='bold', 
            color=obj['color'],
            bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax.grid(True, alpha=0.3)
ax.set_xlabel('X坐标')
ax.set_ylabel('Y坐标')

plt.tight_layout()
plt.show()
```

### 3. 图像分割（Segmentation）
**特点**：像素级的精细分类
**类型**：语义分割、实例分割、全景分割
**优势**：比检测更精确，提供完整的物体边界

```python
# 模拟图像分割可视化
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# 左图：原始图像
ax1.set_title('原始图像', fontsize=14, fontweight='bold')
# 创建一个简单的场景
scene = np.zeros((8, 10, 3))
scene[3:5, 2:4] = [1, 0, 0]  # 红色区域（猫）
scene[1:3, 6:8] = [0, 1, 0]  # 绿色区域（狗）
scene[5:7, 1:3] = [0, 0, 1]  # 蓝色区域（汽车）
ax1.imshow(scene)
ax1.set_xlabel('X坐标')
ax1.set_ylabel('Y坐标')
ax1.grid(True, alpha=0.3)

# 右图：分割结果
ax2.set_title('分割结果', fontsize=14, fontweight='bold')
# 创建分割掩码
mask = np.zeros((8, 10))
mask[3:5, 2:4] = 1  # 猫区域
mask[1:3, 6:8] = 2  # 狗区域
mask[5:7, 1:3] = 3  # 汽车区域
im = ax2.imshow(mask, cmap='tab10', alpha=0.8)
ax2.set_xlabel('X坐标')
ax2.set_ylabel('Y坐标')
ax2.grid(True, alpha=0.3)

# 添加颜色条
cbar = plt.colorbar(im, ax=ax2, fraction=0.046, pad=0.04)
cbar.set_ticks([1, 2, 3])
cbar.set_ticklabels(['猫', '狗', '汽车'])

plt.tight_layout()
plt.show()
```

### 4. 目标追踪（Tracking）
**特点**：在时间维度上跟踪目标
**挑战**：遮挡、相似目标、快速运动
**应用**：视频监控、运动分析、自动驾驶

```python
# 目标追踪轨迹可视化
fig, ax = plt.subplots(1, 1, figsize=(12, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 8)
ax.set_title('目标追踪轨迹', fontsize=14, fontweight='bold')
ax.set_xlabel('X坐标')
ax.set_ylabel('Y坐标')

# 模拟目标在不同帧的位置
frames = [1, 2, 3, 4, 5, 6, 7, 8]
target1_x = [2, 2.5, 3, 3.5, 4, 4.5, 5, 5.5]
target1_y = [3, 3.2, 3.1, 3.3, 3.5, 3.7, 4, 4.2]
target2_x = [7, 6.8, 6.5, 6.2, 5.8, 5.5, 5, 4.5]
target2_y = [2, 2.3, 2.5, 2.7, 2.9, 3.1, 3.3, 3.5]

# 绘制轨迹
ax.plot(target1_x, target1_y, 'r-o', linewidth=2, markersize=6, label='目标1（行人）')
ax.plot(target2_x, target2_y, 'b-o', linewidth=2, markersize=6, label='目标2（车辆）')

# 标记起始和结束点
ax.scatter(target1_x[0], target1_y[0], c='red', s=100, marker='s', label='起点1')
ax.scatter(target1_x[-1], target1_y[-1], c='red', s=100, marker='*', label='终点1')
ax.scatter(target2_x[0], target2_y[0], c='blue', s=100, marker='s', label='起点2')
ax.scatter(target2_x[-1], target2_y[-1], c='blue', s=100, marker='*', label='终点2')

# 添加帧数标记
for i, (x, y) in enumerate(zip(target1_x[::2], target1_y[::2])):
    ax.annotate(f'帧{i*2+1}', (x, y), xytext=(5, 5), textcoords='offset points', fontsize=8)

ax.grid(True, alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```

## 🧪 实践练习

### 练习1：基础概念理解
让我们通过代码来验证对不同概念的理解：

```python
# 模拟不同任务的输入输出
def simulate_cv_task(task_type, input_data):
    """模拟不同计算机视觉任务的输出"""
    if task_type == 'classification':
        return {
            'task': '图像分类',
            'input': '一张图片',
            'output': ['猫', '狗', '汽车'],  # 可能的分类结果
            'confidence': [0.7, 0.2, 0.1]  # 置信度
        }
    elif task_type == 'detection':
        return {
            'task': '目标检测',
            'input': '一张图片',
            'output': [
                {'class': '猫', 'bbox': [100, 100, 200, 200], 'confidence': 0.85},
                {'class': '狗', 'bbox': [250, 150, 350, 250], 'confidence': 0.72}
            ]
        }
    elif task_type == 'segmentation':
        return {
            'task': '图像分割',
            'input': '一张图片',
            'output': {
                'categories': ['背景', '猫', '狗'],
                'mask_size': (224, 224),
                'pixel_counts': [50000, 15000, 10000]
            }
        }
    elif task_type == 'tracking':
        return {
            'task': '目标追踪',
            'input': '视频序列 (30帧)',
            'output': {
                'tracked_objects': 3,
                'trajectory_length': 30,
                'average_speed': [2.5, 1.8, 3.2],  # 像素/帧
                'tracking_accuracy': 0.92
            }
        }

# 测试不同任务
task_results = {}
for task in ['classification', 'detection', 'segmentation', 'tracking']:
    task_results[task] = simulate_cv_task(task, 'sample_input')

# 显示结果
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('计算机视觉任务模拟结果', fontsize=16, fontweight='bold')

task_names = ['图像分类', '目标检测', '图像分割', '目标追踪']
for idx, (task, result) in enumerate(task_results.items()):
    row, col = idx // 2, idx % 2
    ax = axes[row, col]
    
    ax.set_title(f'{task_names[idx]} - 输入输出对比', fontweight='bold')
    
    # 绘制输入输出对比
    x_pos = np.arange(2)
    y_values = [1, len(str(result['output']))]  # 简化的输入输出复杂度
    
    ax.bar(x_pos - 0.2, [1, 1], 0.4, label='输入', alpha=0.7, color='lightblue')
    ax.bar(x_pos + 0.2, [1, y_values[1]], 0.4, label='输出复杂度', alpha=0.7, color='lightcoral')
    
    ax.set_xticks(x_pos)
    ax.set_xticklabels(['输入', '输出'])
    ax.set_ylabel('相对复杂度')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 打印详细的任务信息
print("🔍 各任务详细信息：")
for task, result in task_results.items():
    print(f"\n📋 {result['task']}:")
    print(f"  输入: {result['input']}")
    print(f"  输出: {result['output']}")
    if 'confidence' in result:
        print(f"  置信度: {result['confidence']}")
```

### 练习2：技术演进分析

```python
# 技术演进分析
years = [2012, 2015, 2018, 2021, 2024]
cnn_accuracy = [85, 92, 95, 97, 98]  # CNN在ImageNet上的准确率
vit_accuracy = [0, 0, 88, 94, 97]   # ViT的准确率
yolo_speed = [0, 45, 67, 78, 85]    # YOLO FPS (frames per second)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# 左图：准确率演进
ax1.plot(years, cnn_accuracy, 'b-o', linewidth=2, markersize=8, label='CNN')
ax1.plot(years, vit_accuracy, 'r-s', linewidth=2, markersize=8, label='ViT')
ax1.set_xlabel('年份')
ax1.set_ylabel('准确率 (%)')
ax1.set_title('计算机视觉模型准确率演进')
ax1.set_ylim(80, 100)
ax1.legend()
ax1.grid(True, alpha=0.3)

# 添加关键里程碑
milestones = [
    (2012, 'AlexNet'),
    (2015, 'ResNet'),
    (2018, 'EfficientNet'),
    (2021, 'Vision Transformer'),
    (2024, 'DINOv2')
]

for year, name in milestones:
    ax1.axvline(x=year, color='gray', linestyle='--', alpha=0.5)
    ax1.text(year, 82, name, rotation=45, fontsize=8, ha='right')

# 右图：检测速度演进
ax2.bar(years, yolo_speed, alpha=0.7, color='green', width=0.8)
ax2.set_xlabel('年份')
ax2.set_ylabel('FPS (帧/秒)')
ax2.set_title('YOLO检测速度演进')
ax2.set_ylim(0, 100)
ax2.grid(True, alpha=0.3)

yolo_versions = ['YOLOv1', 'YOLOv2', 'YOLOv3', 'YOLOv5', 'YOLOv8']
for i, (year, fps, version) in enumerate(zip(years, yolo_speed, yolo_versions)):
    ax2.text(year, fps + 2, version, ha='center', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.show()

print("📈 技术演进洞察：")
print("• 准确率：从CNN的85%提升到现在的97%+")
print("• 架构：从传统CNN发展到ViT等Transformer架构")
print("• 速度：YOLO从最初的45FPS提升到现在的85FPS+")
print("• 应用：从学术研究走向大规模商业应用")
```

## 🎓 学习总结与展望

### 核心要点回顾
1. **图像分类**：最基础的任务，回答"这是什么？"
2. **目标检测**：找出"哪里有什么？"，需要定位和分类
3. **图像分割**：最精细的像素级分类，提供精确边界
4. **目标追踪**：在时间维度上跟踪目标运动

### 技术演进趋势
- **架构演进**：CNN → Transformer → 混合架构
- **性能提升**：准确率↑、速度↑、计算效率↑
- **应用深化**：从单任务到多任务协同、从静态到动态
