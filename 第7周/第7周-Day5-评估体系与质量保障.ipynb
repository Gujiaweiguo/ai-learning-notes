```python
# 配置 matplotlib 中文显示
from matplotlib import font_manager
import matplotlib.pyplot as plt
import numpy as np

font_path = "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc"
font_manager.fontManager.addfont(font_path)
font_name = font_manager.FontProperties(fname=font_path).get_name()
plt.rcParams["font.family"] = font_name
plt.rcParams["axes.unicode_minus"] = False
print(f"中文字体配置完成: {font_name}")
```

## 实验1：评估指标雷达图

```python
from math import pi

dimensions = ["准确性", "完整性", "一致性", "及时性", "合规性", "用户满意度"]
N = len(dimensions)
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

v1 = [6, 5, 7, 4, 8, 5, 6]
v2 = [7, 7, 8, 6, 8, 7, 7]
v3 = [9, 8, 9, 8, 9, 8, 9]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))
ax.plot(angles, v1, 'o-', linewidth=1.5, label="v1.0 初始版", color="#FF6B6B")
ax.fill(angles, v1, alpha=0.1, color="#FF6B6B")
ax.plot(angles, v2, 's-', linewidth=1.5, label="v2.0 优化版", color="#4ECDC4")
ax.fill(angles, v2, alpha=0.1, color="#4ECDC4")
ax.plot(angles, v3, '^-', linewidth=2, label="v3.0 成熟版", color="#45B7D1")
ax.fill(angles, v3, alpha=0.1, color="#45B7D1")
ax.set_xticks(angles[:-1])
ax.set_xticklabels(dimensions, fontsize=11)
ax.set_ylim(0, 10)
ax.set_title("数字员工评估雷达图", fontsize=14)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1))
plt.tight_layout()
plt.show()
```

## 实验2：护栏触发率与误杀率

```python
thresholds = np.linspace(0, 1, 50)
trigger_rate = 1 - np.exp(-3 * thresholds)
false_positive = thresholds ** 2 * 0.3

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(thresholds, trigger_rate, '-', color="#FF9800", linewidth=2, label="护栏触发率")
ax.plot(thresholds, false_positive, '-', color="#E91E63", linewidth=2, label="误杀率（误拦正常请求）")
ax.fill_between(thresholds, trigger_rate, false_positive,
                where=(trigger_rate > false_positive), alpha=0.2, color="green", label="有效防护区域")
ax.set_xlabel("护栏阈值（严格程度）")
ax.set_ylabel("比率")
ax.set_title("护栏阈值 vs 触发率/误杀率（模拟）")
ax.legend()
ax.axvline(x=0.35, color='gray', linestyle='--', alpha=0.7)
ax.text(0.36, 0.5, "推荐阈值", fontsize=10, color="gray")
plt.tight_layout()
plt.show()
print("💡 关键洞察：阈值太低→漏放危险请求；太高→误拦正常业务")
```

## 实验3：多版本评估指标趋势

```python
versions = ["v1.0", "v1.1", "v1.2", "v2.0", "v2.1", "v3.0"]
accuracy = [72, 76, 80, 85, 88, 93]
compliance = [85, 88, 88, 90, 92, 95]
latency_p95 = [3.2, 3.0, 2.8, 2.5, 2.3, 2.0]
user_satisfaction = [65, 70, 72, 80, 85, 90]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, data, title, color in [
    (axes[0,0], accuracy, "准确性 (%)", "#4CAF50"),
    (axes[0,1], compliance, "合规率 (%)", "#2196F3"),
    (axes[1,0], latency_p95, "P95延迟 (s)", "#FF9800"),
    (axes[1,1], user_satisfaction, "用户满意度 (%)", "#9C27B0"),
]:
    ax.plot(versions, data, 'o-', color=color, linewidth=2, markersize=6)
    ax.set_title(title)
    ax.tick_params(axis='x', rotation=45)
plt.suptitle("数字员工版本迭代质量趋势", fontsize=14)
plt.tight_layout()
plt.show()
```
