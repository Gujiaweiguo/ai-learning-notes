## 🔁 ReAct框架：Agent的工作流程

ReAct框架是Agent的灵魂，它定义了Agent如何思考和行动：

**Reason → Action → Observation**

1. **思考（Reason）**：分析当前 situation
2. **行动（Action）**：选择合适的工具执行
3. **观察（Observation）**：获取结果并评估

```python
# 模拟Agent的ReAct工作流程
def simulate_agent_workflow():
    """模拟Agent处理"查询今天天气"任务的完整流程"""
    
    # 步骤1：Reason - 分析需求
    print("🤖 步骤1：思考（Reason）")
    print("用户要求：'查询今天天气'")
    print("分析：需要天气工具，参数：位置（默认本地）")
    
    # 步骤2：Action - 执行工具
    print("\n🔧 步骤2：行动（Action）")
    print("调用工具：weather_api.get_current_weather()")
    
    # 步骤3：Observation - 获取结果
    print("\n👁️  步骤3：观察（Observation）")
    print("天气数据：晴天，25°C，微风")
    
    # 步骤4：总结响应
    print("\n✅ 步骤4：生成最终回复")
    print("回复用户：'今天天气晴朗，温度25°C，建议外出！'")
    
    return True

# 运行模拟
simulate_agent_workflow()

# 可视化ReAct循环
fig, ax = plt.subplots(1, 1, figsize=(8, 6))

# 创建循环图
steps = ['Reason\n(思考)', 'Action\n(行动)', 'Observation\n(观察)']
positions = [(0, 1), (1, 0), (1, 2)]
arrows = [(0.2, 0.8), (0.8, 0.2), (0.8, 1.8)]

# 绘制步骤
for i, (step, pos) in enumerate(zip(steps, positions)):
    circle = plt.Circle(pos, 0.3, color=f'C{i}', alpha=0.7)
    ax.add_patch(circle)
    ax.text(pos[0], pos[1], step, ha='center', va='center', fontweight='bold', fontsize=12)

# 绘制箭头
for i in range(3):
    next_i = (i + 1) % 3
    ax.annotate('', xy=positions[next_i], xytext=positions[i],
                arrowprops=dict(arrowstyle='->', lw=2, color='gray'))

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-0.5, 2.5)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('ReAct循环：Agent的工作流程', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 🛠️ Agent vs 传统编程

这是一个非常重要的概念！理解两者的区别才能真正掌握Agent的精髓。

**传统编程**：
📍 程序员写死所有规则
📍 处理已知情况
📍 缺乏灵活性

**Agent编程**：
📍 Agent动态决策
📍 能处理未知情况
📍 具有自适应能力

```python
# 对比：传统编程 vs Agent编程
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# 传统编程
traditional_code = '''
def process_order(order):
    if order.items == 'A':
        return handle_A()
    elif order.items == 'B':
        return handle_B()
    else:
        return "未知订单"
'''

# Agent编程
agent_code = '''
def agent_process_order(order, tools):
    # 分析订单类型
    analysis = llm.analyze(order.description)
    
    # 根据分析选择工具
    if analysis.requires_payment:
        result = tools.process_payment(order)
    elif analysis.requires_inventory:
        result = tools.check_inventory(order)
    else:
        result = tools.handle_unknown(order)
    
    return result
'''

# 可视化对比
ax1.text(0.1, 0.5, traditional_code, fontsize=10, fontfamily='monospace', 
         verticalalignment='center', transform=ax1.transAxes)
ax1.set_title('传统编程：写死规则', fontsize=14, fontweight='bold')
ax1.axis('off')

ax2.text(0.1, 0.5, agent_code, fontsize=10, fontfamily='monospace', 
         verticalalignment='center', transform=ax2.transAxes)
ax2.set_title('Agent编程：动态决策', fontsize=14, fontweight='bold')
ax2.axis('off')

plt.tight_layout()
plt.show()

print('🔍 关键区别：Agent能根据实际情况动态选择策略，而不是依赖预定义的规则！')
```

## 💡 糖水店业务场景应用

让我们想想Jason的糖水店如何利用Agent提升效率：

🔥 **订单处理流程**：
- 自动接收订单
- 分析订单复杂度
- 分配给合适员工
- 跟踪制作进度

📊 **库存管理**：
- 实时监控库存
- 预测需求
- 自动下单采购

👥 **客户服务**：
- 智能回复常见问题
- 个性化推荐
- 客户满意度分析

```python
# 模拟糖水店Agent的工作场景
class DessertShopAgent:
    def __init__(self):
        self.inventory = {
            '红豆': 50, '绿豆': 40, '西米': 30,
            '椰奶': 20, '牛奶': 25, '珍珠': 35
        }
        self.tools = {
            'check_inventory': self.check_inventory,
            'process_order': self.process_order,
            'send_notification': self.send_notification
        }
    
    def check_inventory(self, item):
        """检查库存"""
        return self.inventory.get(item, 0)
    
    def process_order(self, order):
        """处理订单"""
        ingredients = order['ingredients']
        for item, amount in ingredients.items():
            if self.check_inventory(item) < amount:
                return f"库存不足：{item}"
        return "订单可处理"
    
    def send_notification(self, message):
        """发送通知"""
        print(f"📢 通知：{message}")
        return "通知已发送"
    
    def handle_customer_request(self, request):
        """处理客户请求"""
        print(f"🤖 收到客户请求：{request}")
        
        # 分析需求
        if '红豆沙' in request:
            order = {'ingredients': {'红豆': 2, '椰奶': 1}}
            result = self.tools['process_order'](order)
            print(f"🔍 检查结果：{result}")
            
            if result == "订单可处理":
                self.tools['send_notification']("红豆沙订单已确认")
                return "您的红豆沙正在制作中，请稍候！"
            else:
                return "抱歉，红豆沙暂时缺货"
        
        elif '库存' in request:
            return f"当前红豆库存：{self.check_inventory('红豆')}份"
        
        else:
            return "抱歉，我不太理解您的请求"

# 测试糖水店Agent
agent = DessertShopAgent()

# 模拟不同请求
requests = [
    "我要一份红豆沙，谢谢",
    "红豆还有多少库存？",
    "帮我做个双皮奶"
]

print("🧪 糖水店Agent测试场景：")
print("=" * 50)
for i, req in enumerate(requests, 1):
    print(f"\n场景{i}：{req}")
    response = agent.handle_customer_request(req)
    print(f"💬 Agent回复：{response}")
    print("-" * 30)

# 可视化糖水店Agent的工作流程
fig, ax = plt.subplots(1, 1, figsize=(10, 6))

workflow_steps = [
    '客户请求', '需求分析', '库存检查', 
    '订单处理', '状态通知', '完成确认'
]

y_pos = np.arange(len(workflow_steps))
efficiency = [0.8, 0.9, 0.95, 0.85, 0.9, 1.0]

ax.barh(y_pos, efficiency, color='lightcoral', alpha=0.7)
ax.set_yticks(y_pos)
ax.set_yticklabels(workflow_steps)
ax.set_xlabel('处理效率', fontsize=12)
ax.set_title('糖水店Agent工作流程效率分析', fontsize=16, fontweight='bold')
ax.grid(True, alpha=0.3)

# 添加效率标签
for i, eff in enumerate(efficiency):
    ax.text(eff + 0.01, i, f'{eff:.1%}', va='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
