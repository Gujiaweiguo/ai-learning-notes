## 📝 本周复习要点

### Day 1：FFN、LayerNorm与残差连接 🧱
• **FFN（前馈网络）**：两层全连接网络，在注意力层后增加非线性能力
• **LayerNorm**：对每个样本的特征维度进行归一化，稳定训练过程
• **残差连接**：输入直接加到输出，解决深层网络梯度消失问题
• **核心公式**：`y = LayerNorm(x + FFN(Norm(x)))`（Pre-LN结构）

### Day 2：Tokenizer与词嵌入 📚
• **分词原理**：将文本切分为有意义的子词单元（BPE算法）
• **嵌入层**：将token转换为高维向量（768/4096维）
• **位置编码**：注入序列顺序信息（正弦/余弦函数）
• **核心流程**：文本 → tokens → embeddings → position encoding → input

### Day 3：整体架构回顾 🏗️
• **GPT vs BERT**：生成式 vs 预训练模式对比
• **架构差异**：GPT只有decoder，BERT有encoder+decoder
• **任务适应**：GPT适合文本生成，BERT适合理解任务
• **能力边界**：上下文窗口、任务类型、输出形式

### Day 4：代码实战 NumPy实现Self-Attention ⚡
• **核心三步**：Q/K/V计算 → 注意力权重 → 加权求和
• **关键操作**：矩阵乘法、softmax、mask机制
• **50行代码**：从零实现不依赖任何框架
• **验证方法**：手动输入检查输出合理性

````python
# 可视化Transformer核心组件
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('🤖 Transformer 核心组件可视化', fontsize=16, fontweight='bold')

# 1. 注意力机制热力图
tokens = ['我', '爱', '学', '习', 'AI']
attention_matrix = np.array([
    [0.8, 0.1, 0.05, 0.03, 0.02],
    [0.1, 0.7, 0.15, 0.03, 0.02],
    [0.05, 0.1, 0.6, 0.2, 0.05],
    [0.02, 0.05, 0.2, 0.6, 0.13],
    [0.01, 0.02, 0.08, 0.15, 0.74]
])

sns.heatmap(attention_matrix, annot=True, fmt='.2f', cmap='YlOrRd', 
            xticklabels=tokens, yticklabels=tokens, ax=axes[0,0])
axes[0,0].set_title('🎯 Self-Attention 权重分布')
axes[0,0].set_xlabel('目标 Token')
axes[0,0].set_ylabel('查询 Token')

# 2. FFN 结构示意图
ffn_steps = ['输入 (768维)', 'Linear1 → GELU', 'Linear2 (768维)', '输出']
ffn_values = [1, 0.3, 0.3, 1]

axes[0,1].barh(range(len(ffn_steps)), ffn_values, height=0.6, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
axes[0,1].set_yticks(range(len(ffn_steps)))
axes[0,1].set_yticklabels(ffn_steps)
axes[0,1].set_title('🧱 FFN 前馈网络结构')
axes[0,1].set_xlabel('信息流强度')
for i, v in enumerate(ffn_values):
    axes[0,1].text(v + 0.02, i, f'{v:.1f}', va='center')

# 3. LayerNorm 效果对比
np.random.seed(42)
original_data = np.random.randn(100, 768) * 10 + 5
normalized_data = (original_data - np.mean(original_data, axis=1, keepdims=True)) / np.std(original_data, axis=1, keepdims=True)

axes[1,0].hist(original_data.flatten(), bins=50, alpha=0.7, label='原始数据', color='red')
axes[1,0].hist(normalized_data.flatten(), bins=50, alpha=0.7, label='LayerNorm后', color='blue')
axes[1,0].set_title('📊 LayerNorm 归一化效果')
axes[1,0].set_xlabel('数值范围')
axes[1,0].set_ylabel('频率')
axes[1,0].legend()
axes[1,0].set_xlim(-5, 5)

# 4. 残差连接示意图
residual_steps = ['输入', 'F(x)', 'x + F(x)', '输出']
residual_values = [1, 0.5, 1.3, 1]
colors = ['#96CEB4', '#FFD93D', '#6BCF7F', '#4ECDC4']

for i, (step, value, color) in enumerate(zip(residual_steps, residual_values, colors)):
    axes[1,1].bar(i, value, width=0.8, color=color, alpha=0.8)
    axes[1,1].text(i, value + 0.05, step, ha='center', va='bottom', fontweight='bold')
    if i > 0:
        axes[1,1].annotate('', xy=(i-0.4, 1), xytext=(i-1.2, 0.5), 
                          arrowprops=dict(arrowstyle='->', color='black', lw=2))

axes[1,1].set_title('🔄 残差连接流程')
axes[1,1].set_ylabel('信息量')
axes[1,1].set_xlim(-0.5, 3.5)
axes[1,1].set_xticks([])

plt.tight_layout()
plt.show()

print("📈 可视化解读：")
print("• 热力图展示了注意力机制如何让词关注相关词")
print("• FFN通过网络层提取深层特征表示")
print("• LayerNorm将数据归一化到标准分布，稳定训练")
print("• 残差连接帮助信息穿越深层网络")
````

````python
# 实现一个简化的Self-Attention来验证理解
def simplified_self_attention(input_sequence, d_k=64):
    """简化的Self-Attention实现"""
    # 假设我们有 embedding dimension
    batch_size, seq_len, d_model = input_sequence.shape
    
    # 随机初始化Q, K, V权重矩阵
    W_q = np.random.randn(d_model, d_k) * 0.01
    W_k = np.random.randn(d_model, d_k) * 0.01
    W_v = np.random.randn(d_model, d_k) * 0.01
    
    # 计算Q, K, V
    Q = np.dot(input_sequence, W_q)  # (batch, seq_len, d_k)
    K = np.dot(input_sequence, W_k)  # (batch, seq_len, d_k)
    V = np.dot(input_sequence, W_v)  # (batch, seq_len, d_k)
    
    # 计算注意力分数
    scores = np.matmul(Q, K.transpose(0, 2, 1)) / np.sqrt(d_k)  # (batch, seq_len, seq_len)
    
    # Softmax
    def softmax(x, axis=None):
        exp_x = np.exp(x - np.max(x, axis=axis, keepdims=True))
        return exp_x / np.sum(exp_x, axis=axis, keepdims=True)
    
    attention_weights = softmax(scores, axis=-1)
    
    # 加权求和
    output = np.matmul(attention_weights, V)
    
    return output, attention_weights

# 测试一个简单的例子
np.random.seed(123)
test_input = np.random.randn(1, 5, 768)  # batch=1, seq_len=5, d_model=768
output, attn_weights = simplified_self_attention(test_input)

print("🎮 Self-Attention 测试结果：")
print(f"输入形状: {test_input.shape}")
print(f"输出形状: {output.shape}")
print(f"注意力权重形状: {attn_weights.shape}")
print(f"注意力权重矩阵（第一个词关注其他词的程度）：")
print(attn_weights[0])
print("\n✅ 验证：注意力权重每行之和应为1.0")
print(f"每行之和: {np.sum(attn_weights[0], axis=1)}")

# 可视化注意力权重
tokens = ['Token1', 'Token2', 'Token3', 'Token4', 'Token5']
plt.figure(figsize=(8, 6))
sns.heatmap(attn_weights[0], annot=True, fmt='.3f', cmap='Blues', 
            xticklabels=tokens, yticklabels=tokens)
plt.title('🎯 Self-Attention 权重矩阵（简化实现）')
plt.xlabel('Key Tokens')
plt.ylabel('Query Tokens')
plt.show()
````
