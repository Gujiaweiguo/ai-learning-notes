## 1. 今日核心问题：为什么「已经在执行」≠「已被治理」

今天抽了 34 条规则（§5），全部**正在生产代码里天天执行**：`content_hash` 在每次政策提交时冻结内容指纹；`ResolveEffective` 在每次批量定价时判定越权。按传统标准它们不是债务——有 DDL 约束、有测试、有 spec 归档。那 W18「Rule 显式化」到底要显式化什么？

分界线在**可寻址性（addressability）**：一条规则被治理，意味着它能被**指名道姓地引用**——「规则 R-23 在哪、谁改的、哪条 spec 承认它、它约束哪些实体」。今天 34 条语料里，DDL CHECK 和纯函数能通过 grep 找到，但找到的是**代码位置**不是**规则身份**；改一行 `stricterFloor` 的比较符，没有任何门会红，因为它没有指纹、没有注册表条目、没有 citing 关系。更刺眼的证据是今天的**注释级规则漂移首案**：`unit_pricing.go:404-406` 留着 rev1 的 `decideOverAuthority` 注释（「below a project-source floor escalates one level (city); below an HQ/intersect floor escalates to HQ」），而实现早已迁移改名到 `unitpricingbatch/diagnostics.go:171` `decideMinApprovalLevel`（新址注释自称 "mirrors rev1 decideOverAuthority"）——**规则还在执行，规则的「说明书」已经指向空房间**。执行存在 ≠ 治理存在，正如 D3 的结论（描述存在 ≠ 资产有人认领）在规则层的复刻。


## 2. 人话解释

把两链规则翻译成商业地产的话：**招商政策（leasing-policy）= 租决文件的版本控制**。一份租决报批后不许改，要改就复制出新版本重新走审批——「提交时冻结指纹」就是报批那一刻给文件拍照存档，照片上连日期审批人都刻意不拍（生命周期不属于内容）；作废必须写原因且历史永留。**价格授权（unit-pricing）= 集团底价与项目底线的关系**。集团给所有项目划了价格红线，项目只能把线收得更紧（底价更高、上限更低），永远不能放宽——机器判定时取「更严交集」；铺位租金低于项目级红线，审批升一级到城市公司，低于集团红线直接升到集团，只超上限不升级但必须写原因。这两段话是「规则的业务语义」，而它们今天只存在于 Go 代码的 if/else、SQL 的 WHERE 和注释里——W18 要做的不是重写这些 if，是让这段话**有机器可读的身份证**，使「改了比较符」这种事像「改了表名」一样当天现形。

## 6. 商业地产映射

Jason 的 26 年 ERP 经验里这两链都有原型：**租决版本不可变 = 商管公司的「租决文件报批制」**——租决（租赁决策文件）经投委会批准后是合同定价的依据，改一个字都要重新报批；老 ERP（SAP RE-FX / 明源）用「文档版本 + 审批流绑定」实现，痛点从来不是「能不能改」（不能），而是**「事后说不清某份合同当时适用的租决是哪版、内容是什么」**——mi 的 content_hash + 版本绑定把这个审计问题变成了指纹问题（D3 的 A16 命中聚合还能反向回答「这份报价符不符合当时政策」）。**三级价格授权 = 集团-城市-项目的价格权限矩阵**——老 ERP 用审批流条件（release strategy）实现，痛点是**红线本身没有版本**：集团改了底价，历史批次按哪条线判？mi 的约束行自带生效窗（effective_from/to），`ResolveEffective(at)` 按**当时**的约束判**当时**的批次——这是老方案普遍缺的时间旅行能力。而 B8 的「city 一级」硬编码正是老 ERP 的经典病：组织调整（三级改两级）时，审批流条件里的层级引用全部要人肉排查——今天它以 Go 常量的形态复发。


## 7. 与传统方案比较

传统 ERP 的规则形态谱系：①DB 触发器/约束（= 本仓 DDL）②存储过程 if（= 硬编码 service）③工作流引擎条件（= workflow + spec）④配置表（= coding_rules）。**形态上本仓没有发明任何新东西，甚至更朴素**（resolve.go 是纯函数而非规则引擎）。差异在治理面：传统方案的四形态各有日志各有维护界面，但**没有一条规则有全局身份**——审计问「价格红线规则最近一次变更是谁批的」，四形态各答各的。W18 显式化的主张不是换掉四形态（那是规则引擎厂商的推销话术），是照抄 G-07 已验证的「集合 + 基线 + 引用承认」给规则宇宙装门：34 条语料 = 集合的种子，registry = 基线，代码锚点（file:line）= 引用。**规则引擎解决执行一致性，注册表解决治理可寻址性——两个问题，后者更便宜也更没人做**。


## 8. 架构师思考题

1. **哈希里为什么要放 version_no（A4）**？两个版本条款完全一致（copy 后不改任何业务字段直接提交），指纹却不同——业务上这算「同一政策」还是「两个政策」？如果把 version_no 拿掉，A14 的「复制防碰撞」就要另找机制；放进去，「条款相同的不同报批」永远无法去重。这个 19 字注释（hash.go:19-20）里藏着一个业务身份的定义问题——W18 显式化时这条要写成显式决策还是保持隐式？
2. **BELOW_FLOOR 为什么是 flag 不是 block（B7）**？老 ERP 的价格红线是硬约束（低于红线录不进去），mi 设计成软约束（可提交，走特批升级路径）——这背后是招商业务的价格弹性假设：谈判中「低于底价但值得做」的案子真实存在，硬约束会把生意逼到系统外。但软约束的代价是**每一条越权都需要有人在审批流里负责**。什么规则该硬（DDL block）、什么该软（flag + escalate）——这条分界线本身是不是也该是一条显式规则？
3. **B9 的孤儿注释怎么防**？注释不编译、不测试、不 review 拦截——refactor 搬走实现时注释被留下是必然而非事故。t01 的 Go↔registry 描述 diff 模式能防工具描述漂移，但「注释级规则漂移」没有等价物：W18 的规则 registry 要不要把「规则出处注释」也纳入锚点体系（锚到注释行，搬家即 RED）？——G-05 的 expect 签名恰好是为这种「原样存在性」证明设计的。


## 9. 我的理解变化（保留学习期传统）

- **以前以为**：W18 Rule 显式化 = 把 if-else 重写成规则引擎/DSL，是「写新东西」。**现在知道**：第一性问题是**可寻址性**——34 条里 12 条主形态已是 DDL/纯函数引擎（本来就声明式），显式化的主体工作是「给已存在的规则发身份证」（registry + 锚点 + citing），照抄表宇宙已验证的模式。真正要「发明」的集中在高危 5 条（B3/B8/B14 跨处控制流 + A3/B9 注释契约）。
- **以前以为**：注释是代码里最无害的部分，写清楚是加分项。**现在知道**：**注释是规则的第六种形态，而且是最脆弱的一种**——它不编译、不测试、不拦 review，B9 首案里规则的真身搬了家，说明书留在原址继续「教」下一个人。D3 说「描述是 AI 的接口契约」，今天补下半句：**注释是规则的自认书，而自认书会说谎**。
- **以前以为**：探针是「跑个脚本看数字」，环境稳定是当然前提。**现在知道**：S2 日探针的可靠性锚在「本地三仓副本存在」这个未声明的前提上——沙箱里副本消失，脚本第一行就 FATAL。修复方式仍是 D1 的老律：**人肉恢复（手动 fetch 重算）是补丁，把「探针可重建」写进 probe.py（无本地仓时自动 clone/fetch 模式）才是把前提变成代码**。今日方法注记即补丁的诚实记录。
