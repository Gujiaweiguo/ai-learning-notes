# W11-D7 概念实验：Virtual CTO 总复盘

核心概念来自 Markdown：理解深度从 W8 到 W11 上升，但五维工程评分因测量精度提高而下探。下面重现“剪刀差”，并计算 Charter 状态与路线图优先级。

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import font_manager

font_path = "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc"
font_manager.fontManager.addfont(font_path)
font_name = font_manager.FontProperties(fname=font_path).get_name()
plt.rcParams["font.family"] = font_name
plt.rcParams["axes.unicode_minus"] = False
```

```python
weeks = np.array([8, 9, 10, 11])
understanding = np.array([7.0, 7.5, 8.5, 9.0])
engineering = np.array([7.2, 6.8, 6.8, 6.4])
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(weeks, understanding, "o-", linewidth=2, label="理解深度")
ax.plot(weeks, engineering, "o-", linewidth=2, label="五维工程评分")
ax.set_xticks(weeks, [f"W{w}" for w in weeks])
ax.set_ylim(5.5, 10)
ax.set_ylabel("评分")
ax.set_title("从望远镜到内窥镜：理解上升与工程评分下探")
ax.grid(alpha=0.25)
ax.legend()
plt.tight_layout()
plt.show()
plt.close(fig)
print("理解增幅:", understanding[-1] - understanding[0], "工程评分变化:", engineering[-1] - engineering[0])
```

```python
charter = {"FrozenExecutionContext": "已实现", "Canonical Path": "已实现", "Catalog Projection": "已实现", "Knowledge Closure": "部分", "RuntimeABI": "目标态", "Deterministic Build": "未完成", "Promotion Audit": "部分", "PII Default": "未修复"}
from collections import Counter
counts = Counter(charter.values())
print("Charter 状态计数:", dict(counts))
for status, count in counts.items():
    print(f"{status}: {count}/{len(charter)}")
```

```python
priorities = [
    ("PII 默认开启", 5, "安全"),
    ("KnowledgeSnapshot", 5, "可复现闭包"),
    ("RuntimeABI", 5, "版本兼容"),
    ("CapabilityRelease", 4, "发布治理"),
    ("Connector 统一抽象", 3, "外部连接"),
]
priority_scores = np.array([item[1] for item in priorities])
fig, ax = plt.subplots(figsize=(8, 3.8))
ax.barh([item[0] for item in priorities][::-1], priority_scores[::-1], color="#4c78a8")
ax.set_xlim(0, 5.5)
ax.set_xlabel("优先级（5=最高）")
ax.set_title("复盘后行动优先级")
for y, item in enumerate(priorities[::-1]):
    ax.text(item[1] + 0.08, y, item[2], va="center")
plt.tight_layout()
plt.show()
plt.close(fig)
print("结论：评分下行本身不是系统变差的证据；它暴露了闭包、版本、PII 和连接治理等应进入 Sprint backlog 的事实。")
```
