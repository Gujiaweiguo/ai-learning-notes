## 1. DDD 战术设计三件套（60秒版）

### ① Entity（实体）
> 有唯一身份标识（Identity），身份不随属性变化而改变。
- 有 ID——即使所有属性都变了，它还是它
- 有生命周期——会创建、修改、销毁
- **MI 例子**：`Contract`（合同）

### ② Value Object（值对象）
> 没有身份标识，完全由属性值决定相等性。不可变。
- 无 ID——两个值对象属性相同就是"相等"
- 不可变——创建后不修改，要改就创建新的
- **MI 例子**：`PhysicalAttributes`（面积、工程条件）

### ③ Aggregate（聚合）
> 一组紧密关联的 Entity 和 Value Object 的集合，作为一个一致性边界统一操作。
- **Aggregate Root**：唯一的外部入口
- **Invariant（不变量）**：聚合内部必须始终保持一致的业务规则
- **事务边界**：一次操作只修改一个 Aggregate
- **MI 例子**：`Asset Foundation` 聚合

## 4. Aggregate 边界 = AI Agent 的认知边界

这是今天最重要的认知。

**传统视角**：Aggregate 是事务一致性边界
> 一次操作只修改一个 Aggregate，保证数据一致性。

**Ontology 视角**：Aggregate 是 AI Agent 的推理边界
> Agent 在一个 Aggregate 内部可以安全推理（不变量保证语义一致性）。跨 Aggregate 推理需要通过显式声明的 effect 关系。

| 推理场景 | 在一个聚合内 | 跨聚合 |
|---|---|---|
| "这个铺位面积多少？" | ✅ Asset Foundation 内 | 不需要跨 |
| "这个铺位能出租吗？" | ❌ 需要 Lease/Occupancy 的规则 | 跨 Asset + Lease |
| "合同终止后影响什么？" | ❌ 需要 effect-registry | 跨 Contract + Lease + Billing |

**结论**：Aggregate 边界 = Agent 安全推理范围。effect-registry = 跨边界推理的"桥梁说明"。


```python
# effect-registry 从聚合间因果关系看
fig, ax = plt.subplots(figsize=(14, 5))
ax.axis('off')

effects = [
    ('occupancy-effect', 'Contract 聚合', 'Asset + Lease 聚合', '合同生命周期 → 资源占用状态变更', '#E91E63'),
    ('financial-effect', 'Contract 聚合', 'Billing 聚合', '合同事件 → 账单/应收生成', '#FF9800'),
    ('state-transition-effect', '任意聚合', '任意聚合', '对象状态传播（通用型）', '#607D8B'),
    ('lead-conversion-effect', 'Leasing Pipeline', 'Leasing Pipeline 内部', '招商阶段转化', '#4CAF50'),
    ('maintenance-effect', 'Operations 聚合', 'Work Order 聚合', '巡检异常 → 工单生成', '#2196F3'),
]

for i, (name, src, dst, desc, color) in enumerate(effects):
    y = 4 - i * 0.9
    ax.fill_between([0.3, 13.5], y-0.3, y+0.3, color=color, alpha=0.08)
    ax.text(1.8, y, name, fontsize=9, fontweight='bold', ha='center', va='center', color=color)
    ax.text(4.5, y, src, fontsize=8, ha='center', va='center')
    ax.text(6, y, '→', fontsize=12, ha='center', va='center', fontweight='bold', color=color)
    ax.text(8.0, y, dst, fontsize=8, ha='center', va='center')
    ax.text(11.5, y, desc, fontsize=7.5, ha='center', va='center', color='#555')

# Header
ax.fill_between([0.3, 13.5], 4.7, 5.1, color='#37474F', alpha=0.9)
ax.text(1.8, 4.9, 'Effect 类型', fontsize=9, fontweight='bold', ha='center', va='center', color='white')
ax.text(4.5, 4.9, '触发方', fontsize=9, fontweight='bold', ha='center', va='center', color='white')
ax.text(8.0, 4.9, '被影响方', fontsize=9, fontweight='bold', ha='center', va='center', color='white')
ax.text(11.5, 4.9, '语义', fontsize=9, fontweight='bold', ha='center', va='center', color='white')

ax.set_xlim(0, 14)
ax.set_ylim(-1, 5.5)
ax.set_title('effect-registry：跨聚合因果关系的显式声明（AI Agent 的推理桥梁）', fontsize=12, fontweight='bold', pad=15)
ax.text(7, -0.5, '⚠️ occupancy-effect 和 financial-effect 是目前最重要的跨聚合因果关系',
        ha='center', fontsize=8, color='red', style='italic')

plt.tight_layout()
plt.savefig('/tmp/pt_w1_d3_effects.png', dpi=150, bbox_inches='tight')
plt.show()
print("effect关系图已生成")
```

## ✏️ 练习（5分钟）

> 看看 Occupancy 聚合。`AvailableForLeasing` 是组合判断：
> Asset.Active ∧ no Occupancy ∧ no Restriction
>
> 如果 AI Agent 要回答"A101 能不能出租？"，它需要分别查 Asset Foundation 聚合和 Lease/Occupancy 聚合，然后组合判断。
>
> **思考**：这个"组合判断"逻辑应该放在哪里？
> - A. 写在 Agent 的 prompt 里？
> - B. 声明在 Ontology 的 Rule 层？
> - C. 编码在某个聚合的方法里？
>
> 提示：Agent 不可靠的地方，就是你需要显式声明的地方。

**答案倾向**：B（Rule 层）。因为 Agent prompt 不可靠（可能被忽略），聚合方法不可见（Agent 读不到代码）。Rule 层是 AI Agent 能读取并遵守的显式声明。
