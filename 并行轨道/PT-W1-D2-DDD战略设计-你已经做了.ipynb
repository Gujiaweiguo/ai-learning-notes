## 1. DDD 战略设计三大原则

### ① Bounded Context 划分
> 每个 Context 应该有明确的业务边界，内部有一致的语义模型。

检验标准：一个概念在不同 Context 里可以有不同的含义。

**例子**：`Space`（空间）
- Space Management：铺位的物理属性（面积、楼层、位置）
- Lease Management：可租赁单元（租金、租期、状态）
- Property Service：维护对象（设备、工单、巡检）

### ② Context Map（上下文映射）
> Context 之间如何协作？类型：
- **Shared Kernel**（共享内核）
- **Customer-Supplier**（客户-供应商）
- **Conformist**（遵奉者）
- **Anti-Corruption Layer**（防腐层）

### ③ Core Domain vs Supporting Domain
> 什么是你最核心的竞争力？什么是支撑性的？

## 3. 你的 Context Map 现状

| 关系类型 | 示例 | 你的现状 |
|---------|------|---------|
| Customer-Supplier | Billing → Payment | ✅ 隐式存在 |
| Shared Kernel | Lease 和 Contract 共享 Lease 概念 | ⚠️ 没有显式声明 |
| ACL | Payment 不应直接依赖 Lease 内部状态 | ❌ 可能存在耦合 |

**关键 Gap**：你的 Context 之间的关系是隐式的（代码里调），不是显式的（Context Map 声明）。


```python
# MI 显式 Context Map（Ontology 视角）
fig, ax = plt.subplots(figsize=(14, 8))
ax.axis('off')
ax.set_xlim(0, 14)
ax.set_ylim(0, 10)

nodes = {
    'Customer\nMgmt': (2, 8, '#1565C0'),
    'Lease\nMgmt': (5, 8, '#2E7D32'),
    'Contract\nMgmt': (8, 8, '#E65100'),
    'Tenant\nMgmt': (2, 5, '#1565C0'),
    'Space\nMgmt': (5, 5, '#2E7D32'),
    'Billing\nMgmt': (8, 5, '#E65100'),
    'Payment\nMgmt': (11, 5, '#7B1FA2'),
    'Finance\nSettlement': (11, 8, '#7B1FA2'),
    'Property\nMgmt': (5, 2, '#00838F'),
}

edges = [
    ('Customer\nMgmt', 'Tenant\nMgmt', 'converts to'),
    ('Lease\nMgmt', 'Contract\nMgmt', 'signs'),
    ('Lease\nMgmt', 'Space\nMgmt', 'occupies'),
    ('Contract\nMgmt', 'Billing\nMgmt', 'triggers'),
    ('Billing\nMgmt', 'Payment\nMgmt', 'drives'),
    ('Payment\nMgmt', 'Finance\nSettlement', 'settles'),
    ('Property\nMgmt', 'Space\nMgmt', 'owns'),
]

# Draw nodes
for name, (x, y, color) in nodes.items():
    rect = plt.Rectangle((x-1.1, y-0.5), 2.2, 1.0, facecolor=color, alpha=0.12, edgecolor=color, linewidth=2, rounded=True)
    ax.add_patch(rect)
    ax.text(x, y, name, ha='center', va='center', fontsize=8, fontweight='bold', color=color)

# Draw edges
for src, dst, label in edges:
    sx, sy, _ = nodes[src]
    dx, dy, _ = nodes[dst]
    ax.annotate('', xy=(dx, dy+0.6 if dy < sy else dy-0.6), 
                xytext=(sx, sy-0.6 if dy < sy else sy+0.6),
                arrowprops=dict(arrowstyle='->', color='#666', lw=1.5))
    mx, my = (sx+dx)/2, (sy+dy)/2
    ax.text(mx+0.2, my, label, fontsize=6.5, color='#888', style='italic',
            ha='center', va='center')

ax.set_title('MI Context Map（显式声明版）— AI Agent 的业务地图', fontsize=13, fontweight='bold', pad=15)
ax.text(7, 0.5, '⚠️ 当前状态：这些关系存在于代码中，但未显式声明。Ontology 要把它们变成 AI 可读的声明。',
        ha='center', fontsize=8, color='red', style='italic')

plt.tight_layout()
plt.savefig('/tmp/pt_w1_d2_context_map.png', dpi=150, bbox_inches='tight')
plt.show()
print("Context Map 已生成")
```
