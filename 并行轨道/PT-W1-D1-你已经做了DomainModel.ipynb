## 3. Ontology 视角：你的 Domain Model 缺什么

| Ontology 维度 | 你有吗 | 在哪里 | 缺什么 |
|--------------|--------|--------|--------|
| **Concept（概念）** | ✅ | 17 Context + 每个内的对象 | 缺业务定义（只有技术定义） |
| **Identity（身份）** | ⚠️ | 表 ID / 编码 | 缺业务身份层级 |
| **Relationship（关系）** | ⚠️ | ADR-006 四类关系 | 缺语义命名（occupies/signs/governs） |
| **Lifecycle（生命周期）** | ⚠️ | status 字段 + effect-registry | 缺完整状态机 + Event 声明 |
| **Rule（规则）** | ⚠️ | 部分代码 + BCM 规则 | 未显式化（AI 不可推理） |
| **Capability（能力）** | ⚠️ | CRE BCM capability 行 | 缺 Skill 映射 |

**关键认知**：你做的是"软件工程师视角的 Domain Model"，缺的是"AI 可理解的语义层"。


```python
# Domain Model vs Ontology 视角对比
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Domain Model 视角
ax = axes[0]
ax.set_xlim(0, 6); ax.set_ylim(0, 8); ax.axis('off')
ax.set_title('Domain Model 视角\n（给工程师看的）', fontsize=11, fontweight='bold', color='#1565C0')
fields = [
    'lease.id',
    'lease.space_id FK → space.id', 
    'lease.tenant_id FK → tenant.id',
    'lease.status = active/terminated',
    'lease.start_date / end_date',
    'lease.rent_amount'
]
for i, f in enumerate(fields):
    ax.text(3, 7-i*0.9, f, ha='center', fontsize=9, fontfamily='monospace', color='#333')
    ax.text(0.3, 7-i*0.9, '📋', fontsize=10)

# Ontology 视角
ax = axes[1]
ax.set_xlim(0, 6); ax.set_ylim(0, 8); ax.axis('off')
ax.set_title('Ontology 视角\n（给 AI Agent 看的）', fontsize=11, fontweight='bold', color='#2E7D32')
fields = [
    'Lease 是租户与空间的法律协议',
    'Lease **occupies** Space（占用关系）',
    'Lease 处于生效期 → Space 被占用',
    'Lease 生效 → Billing 应该产生',
    'Lease 终止 → Space 释放（需inspection）',
    'Lease = 营运收入的法律基础'
]
for i, f in enumerate(fields):
    ax.text(3, 7-i*0.9, f, ha='center', fontsize=8.5, color='#333')
    ax.text(0.3, 7-i*0.9, '🧠', fontsize=10)

plt.suptitle('同一事物，两种描述', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('/tmp/pt_w1_d1_vs.png', dpi=150, bbox_inches='tight')
plt.show()
print("视角对比图已生成")
```

## 4. 为什么 Domain Model ≠ Ontology

- **Domain Model** = 建筑施工图（给工程师看的，关注怎么实现）
- **Ontology** = 建筑说明书（给使用者/AI 看的，关注这是什么、为什么存在、如何使用）

**区别核心**：Domain Model 描述"数据怎么存"，Ontology 描述"业务世界怎么运作"。
